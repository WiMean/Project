```python
# 1. Подключение Google Drive
from google.colab import drive
drive.mount('/content/drive')

# 2. Установка библиотек (2 минуты)
!pip install transformers datasets torch scikit-learn accelerate -q

# 3. Импорты
import pandas as pd
from transformers import MobileBertTokenizer, MobileBertForSequenceClassification, Trainer, TrainingArguments
from datasets import Dataset
import torch
from sklearn.metrics import accuracy_score

print("✅ Google Drive подключен!")
```

```python
# 4. Загрузка твоих ЧИСТЫХ данных
dataset_path = '/content/drive/MyDrive/Mobile_Project/Dataset NewsMiner'

train_df = pd.read_csv(f'{dataset_path}/train_clean.csv')
test_df = pd.read_csv(f'{dataset_path}/test_clean.csv')

print("📊 ТВОИ ДАННЫЕ:")
print(f"Обучаем: {len(train_df)} строк")
print(f"Тест: {len(test_df)} строк")
print("\nПримеры:")
print(train_df[['text', 'label']].head())
```

```python
# 5. Подготовка для MobileBERT (токенизация)
label2id = {'политика': 0, 'экономика': 1, 'наука': 2, 'другие': 3, 'спорт': 4}
id2label = {v: k for k, v in label2id.items()}

# Конвертируем лейблы в числа
train_df['label'] = train_df['label'].map(label2id)
test_df['label'] = test_df['label'].map(label2id)

# Создаем HuggingFace Datasets
train_dataset = Dataset.from_pandas(train_df[['text', 'label']])
test_dataset = Dataset.from_pandas(test_df[['text', 'label']])

# Загружаем MobileBERT
tokenizer = MobileBertTokenizer.from_pretrained('google/mobilebert-uncased')
model = MobileBertForSequenceClassification.from_pretrained(
    'google/mobilebert-uncased',
    num_labels=5,  # 5 категорий max
    id2label=id2label,
    label2id=label2id
)

# Токенизация
def tokenize_function(examples):
    return tokenizer(examples['text'], padding='max_length', truncation=True, max_length=128)

train_dataset = train_dataset.map(tokenize_function, batched=True)
test_dataset = test_dataset.map(tokenize_function, batched=True)

print("✅ Данные токенизированы!")
print(f"Пример токенов: {train_dataset[0]['input_ids'][:10]}...")
```

```python
# 6. ОБУЧЕНИЕ MobileBERT (15 минут на GPU!)
training_args = TrainingArguments(
    output_dir='/content/newsminer-model',
    num_train_epochs=3,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    warmup_steps=50,
    weight_decay=0.01,
    logging_steps=10,
    eval_strategy="epoch",           # ← ИЗМЕНИЛ: было evaluation_strategy
    save_strategy="epoch",
    load_best_model_at_end=True,
    report_to=[],                    # Отключаем wandb
    dataloader_pin_memory=False      # Для стабильности в Colab
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    tokenizer=tokenizer
)

print("🚀 НАЧИНАЕМ ОБУЧЕНИЕ...")
trainer.train()
trainer.save_model('/content/newsminer-mobilebert-final')
tokenizer.save_pretrained('/content/newsminer-mobilebert-final')
print("✅ ✅ ✅ МОДЕЛЬ ОБУЧЕНА И СОХРАНЕНА!")
```

```python
# 7. ПРОВЕРКА КАЧЕСТВА
predictions = trainer.predict(test_dataset)
preds = predictions.predictions.argmax(-1)
accuracy = accuracy_score(test_dataset['label'], preds)

print(f"🎯 ТОЧНОСТЬ: {accuracy:.2%}")
print("\n📈 Примеры предсказаний:")
for i in range(5):
    text = test_df.iloc[i]['text'][:80] + "..."
    true_label = id2label[test_dataset[i]['label']]
    pred_label = id2label[preds[i]]
    status = "✅" if true_label == pred_label else "❌"
    print(f"{status} {text} | Истинная: {true_label} | Предсказание: {pred_label}")
```
